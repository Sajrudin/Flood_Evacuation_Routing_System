```python
import pandas as pd
import requests
import time
import random
```

```python
API_KEY = "9d63c7067fa16abac115433464ccfbd2"

data = []
```

```python
# Dehradun bounding box (lat, lon range)
lat_range = [30.25, 30.45]
lon_range = [77.95, 78.15]

step = 0.02  # controls number of points

lat = lat_range[0]
while lat <= lat_range[1]:
    lon = lon_range[0]
    while lon <= lon_range[1]:

        url = f"https://api.openweathermap.org/data/2.5/weather?lat={lat}&lon={lon}&appid={API_KEY}"

        try:
            res = requests.get(url).json()

            rainfall = res.get("rain", {}).get("1h", 0)
            temp = res["main"]["temp"]
            humidity = res["main"]["humidity"]

            data.append({
                "lat": lat,
                "lon": lon,
                "rainfall": rainfall,
                "temp": temp,
                "humidity": humidity
            })

            print(f"Fetched: {lat}, {lon}")

        except:
            print(f"Error at: {lat}, {lon}")

        time.sleep(1)  # avoid rate limit

        lon += step
    lat += step
```

```python
# Save to CSV
df = pd.DataFrame(data)
df.to_csv("dehradun_weather.csv", index=False)

print("CSV created: dehradun_weather.csv")
```
